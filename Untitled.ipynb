```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

output_root = Path("outputs/group3")
casos = ["CTRL", "SEM_NUC", "SEM_DEP", "SEM_CONG_NUV", "SEM_CONG_CHUVA", "SEM_RIMING", "SEM_HM", "SEM_GELO_NEVE"]

plt.figure(figsize=(10, 6))
for caso in casos:
    caminho = output_root / caso / f"resultados_{caso}.npz"
    if not caminho.exists():
        continue
    data = np.load(caminho, allow_pickle=True)
    t_min = data['t_s'] / 60.0  # converte para minutos
    w_max = np.max(data['w'], axis=(1, 2))  # máximo sobre x e z
    plt.plot(t_min, w_max, label=caso)

plt.xlabel("Tempo (min)")
plt.ylabel("w_max (m/s)")
plt.legend()
plt.grid(alpha=0.3)
plt.title("Comparação da intensidade do updraft – Grupo 3")
plt.savefig("comparacao_wmax_grupo3.png", dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load('outputs/group3/CTRL/resultados_CTRL.npz', allow_pickle=True)
z = data['z_m']          # altura (m)
ix_centro = data['x_m'].shape[0] // 2   # coluna central
ultimo_tempo = -1

w_perfil = data['w'][ultimo_tempo, ix_centro, :]
plt.plot(w_perfil, z)
plt.xlabel("w (m/s)")
plt.ylabel("altura (m)")
plt.title("Perfil de w no centro do domínio, t final")
plt.show()
```
